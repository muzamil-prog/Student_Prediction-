```python
data['Sunshine_hours']=data['Sunshine_hours'].fillna(data['Sunshine_hours'].median())
data['Evaporation_mm'] = data['Evaporation_mm'].fillna(data['Evaporation_mm'].median())
```

```python
data.isnull().sum()
```

```python
# 1. Date ko proper format mein convert karein
data['Date'] = pd.to_datetime(data['Date'])

# 2. Alag alag columns extract karein
data['Year'] = data['Date'].dt.year
data['Month'] = data['Date'].dt.month
data['Day'] = data['Date'].dt.day

# 3. Original Date column ko drop kar dein
data = data.drop(columns=['Date'])
```

```python
data['Country'].unique()
```

```python
data['Location'].unique()
```

```python
 #1. Pehle RainTomorrow ko pakka text se numeric 0 aur 1 mein convert karein
data['RainTomorrow'] = data['RainTomorrow'].map({'Yes': 1, 'No': 0})

# 1. Location aur Country ka group-wise mean nikal kar direct map kar dein
data['Location'] = data['Location'].map(data.groupby('Location')['RainTomorrow'].mean())
data['Country'] = data['Country'].map(data.groupby('Country')['RainTomorrow'].mean())
```

```python

data.head()
```

```python
# Apply One-Hot Encoding to ClimateZone to turn it into clean numeric columns
data = pd.get_dummies(data, columns=['ClimateZone'], drop_first=True)
```

```python
data.head()
```

```python

plt.figure(figsize=(10, 6))

# Scatter plot: X par MinTemp, Y par MaxTemp, aur color (hue) par MeanTemp lagaya hai
sns.scatterplot(
    data=data, 
    x='MinTemp_C', 
    y='MaxTemp_C', 
    hue='MeanTemp_C', 
    palette='coolwarm', 
    alpha=0.7
)

plt.title('Scatter Plot: Relationship between Min, Max, and Mean Temperature')
plt.xlabel('Minimum Temperature (°C)')
plt.ylabel('Maximum Temperature (°C)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title='Mean Temp (°C)')
plt.show()
```
